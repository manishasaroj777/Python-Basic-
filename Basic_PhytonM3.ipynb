5. What is mutability in Python?

  ->Mutability refers to whether an object can be changed after it's created.

      ->Mutable = Can be changed (e.g., list)

      ->Immutable = Cannot be changed (e.g., tuple, string)

6. Why are lists mutable, but tuples are immutable?

  -> Lists allow changes because they are designed to hold varying data.

  ->Tuples are used for fixed collections, so Python keeps them immutable for safety and efficiency.   

7. What is the difference between == and is operators in Python?   
  
  ->  == checks value equality (whether the values are the same)

  ->  is checks identity (whether they point to the same memory location)

8. What are logical operators in Python?
  
   ->Logical operators are used to combine conditional statements:

      and → True if both conditions are true

      or → True if at least one condition is true

      not → Reverses the condition   

9. What is type casting in Python?
   
   -> Type casting is converting one data type into another.
      Example: int("5") → converts string "5" to integer 5

10. What is the difference between implicit and explicit type casting?
   
   -> Implicit casting: Done automatically by Python.
   for eg: a = 5
           b = 2.0
          c = a + b  # Python converts 5 to 5.0    

   -> Explicit casting: Done manually by the programmer.
   for eg: a = int("10")

11. What is the purpose of conditional statements in Python?
   
   -> Conditional statements allow a program to make decisions and execute   certain blocks of code depending on conditions.
   Examples: if, else, elif  

12. How does the elif statement work?
  
  ->elif (else-if) checks another condition if the previous if condition was False. It's used to handle multiple conditions.  

  for eg: if x == 1:
          print("One")
         elif x == 2:
          print("Two")
         else:
          print("Other")

13. What is the difference between for and while loops?
  
  ->for loop: Used when number of iterations is known.
  ->while loop: Used when you want to repeat until a condition is False.   

14. Describe a scenario where a while loop is more suitable than a for loop.

     for example: user_input = ""
                  while user_input != "yes":

```python
# 5. Function to show mutable vs immutable behavior
def modify_data(num, my_list):
    num += 10  # Immutable
    my_list.append(100)  # Mutable

    number = 5
    lst = [1, 2, 3]
    modify_data(number, lst)

    print("Number after function call:", number)      # No change
    print("List after function call:", lst)           # List modified
```

```python
# 14. Reverse a string using while loop
text = input("Enter a string: ")
reversed_text = ""
i = len(text) - 1

while i >= 0:
    reversed_text += text[i]
    i -= 1

print("Reversed string:", reversed_text)
```

```python
# 15. Factorial using while loop

num = int(input("Enter a number: "))
factorial = 1
i = 1

while i <= num:
    factorial *= i
    i += 1

print(f"Factorial of {num} is", factorial)
```
